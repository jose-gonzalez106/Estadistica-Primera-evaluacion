```python
# Para construir el gráfico primero agrupamos y contamos
cuenta_institucion = df.groupby('TIPO DE INSTITUCION').size()

# Frecuencia relativa
tabla_institucion = pd.DataFrame({
    'Frecuencia relativa (h%)': cuenta_institucion / 34372 * 100})

tabla_institucion
```

```python

# Ingresamos las variables
categorias_institucion = tabla_institucion.index

h_institucion = tabla_institucion['Frecuencia relativa (h%)']

# Gráfico circular
fig, ax = plt.subplots(figsize=(8, 6))

circular_institucion = ax.pie(
    h_institucion,
    labels=categorias_institucion,
    autopct='%.2f%%'
)

# Personalización
ax.set_title('Distribución de estudiantes por tipo de institución')

plt.show()
```
